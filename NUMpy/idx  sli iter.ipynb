indexing using boolean arrays

```python
arr=np.random.randint(low=1,high=100,size=20).reshape(4,5)
print(arr)
print()
#find numbers that are greater than 50
print(arr>50)
print()
print(arr[arr>50]) # indexing using boolean arrays 
print()
#FIND ARRAYS THAT ARE GREATER THAN 50 AND ALSO ODD
print(arr[(arr>50) & (arr%2!=0)])

print()
#want to do 0 the numbers that are greater than 50 and are odd
arr[(arr > 50) & (arr % 2 != 0)] = 0
print(arr)
```

BROADCASTING

```python
#scenario 1
a1=np.arange(8).reshape(2,4)
a2=np.arange(8,16).reshape(2,4)

print(a1)
print()
print(a2)
```

```python
#SCENARIO 2
a1=np.arange(9).reshape(3,3)
a2=np.arange(3).reshape(1,3)

print(a1,a2)
print()
print(a1+a2) #this is called broadcasting - it is the ability two apply mathemetical rule of disimilar dimension arrays 
```

```python
arr=np.arange(3).reshape(1,3)
a2=np.arange(12).reshape(4,3)

print(arr)
print()
print(a2)
print()
#if one of the elements of reshape is one then if corresponding of that 1 is same to the other than it can perform operations 

print(arr+a2)
```
