```python
#WAF to print the elements of a list in a single line (list is the parameter)
cities=["delhi", "bhopal", "sehore","pune","gurgaon"]
heroes=["thor", "ironman", "captain americ","shaktiman","toonpur ka superhero"]

print(heroes[0],end=" ")
print(cities, end=" " )#by using end we can print both lines in a single line 
print(heroes)


def print_list(list): #here we makes the function
    for item in list:
        print(item, end=" ")


print_list(cities)
print(cities[0:2])

```

```python
#WAF to find factorial of n (n is the pareameter)
def cal_factorial(n):
    fact=1
    for i in range (1,n+1):
        fact*=i
    print(fact)
        
cal_factorial(3)
cal_factorial(7)
```

```python
def fact(n):
    if(n==0 or n==1):
      return 1
    return fact(n-1)*n
print(fact(3))

```

```python
#sum of first n natural numbers 
def calc_sum(n):
    if (n==0):
        print(n)
    return  calc_sum+(n-1)
  
sum= calc_sum(5)
    
print(sum)
```

```python
#factorial (7)=7*6*5*4*3*2*1 using recurrsion
def factorial(n):
    if (n==0 or n==1):
        return 1
    else:
        return n * factorial(n-1)
    


print(factorial(3))
```
