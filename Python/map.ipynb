```python
import numpy as np
import matplotlib.pyplot as plt

# Given data
m = 0.35          # kg
k = 1.30e2        # N/m
A = 0.11          # m (11 cm)

# Displacement values
x = np.linspace(-A, A, 200)

# Energies
PE = 0.5 * k * x**2
KE = 0.5 * k * (A**2 - x**2)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, KE, label='Kinetic Energy (KE)')
plt.plot(x, PE, label='Potential Energy (PE)')

plt.xlabel("Displacement (m)")
plt.ylabel("Energy (J)")
plt.title("KE vs Displacement and PE vs Displacement")
plt.grid(True)
plt.legend()

plt.show()
```
