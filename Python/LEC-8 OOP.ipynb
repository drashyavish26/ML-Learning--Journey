```python
#PRIVATE ATTRIBUTES
class account :
    def __init__(self,acc_no,acc_pass):
        self.acc_no=acc_no
        self.__acc_pass=acc_pass #we use 2 underscores before an attribute to make it private
   
    def reset_pass(self):
        print(self.__acc_pass) #we can access this attributes inside the class 


acc1 = account("2332432432","drashya")

print(acc1.acc_no)
print(acc1.reset_pass())
print(acc1.acc_pass)

```

INHERITANCE 

```python
class car:  
    @staticmethod
    def start():
        print("car started")

    @staticmethod
    def stop():
        print("car stopped")


class toyotacar(car):
    def __init__(self,name):
        self.name=name

car1=toyotacar("fortuner")
car2=toyotacar("innova")
print(car1.start())


class fortuner(toyotacar): #MULTILEVEL INHERITANCE 
    def __init__(self,type):
        self.type=type

car11=fortuner("diesel")
car11.start()


```

```python
class car:
    def __init__(self,type):
        self.type=type

    @staticmethod #Without @staticmethod, Python expects self as the first parameter.
    def start():
            print("car is started")

    @staticmethod ## Static method tab use karte hai jab function ko object ki information ki zarurat nahi hoti.
    def stop():
            print("car is stopped.....")

class toyotacar(car):
    def __init__(self,name,type):
        super().__init__(type)
        self.name=name
        super().start() 

car1=toyotacar("prius","electric")
print(car1.type)
```
