```python
#BREAK AND CONTINUE
i = 0
while i <=5:
    print(i)
    if(i==3):
        break #it breaks the function at a define point
    i+=1
print("end of loop")


```

```python
#CONTINUE - terminates the current exuction 
i=0
while i<=5:
    if (i==3):
        i+=1
        continue #skips (the steps for this iteration)
    print(i)
    i+=1
```

```python
#WAP to find the factorial of first n natural numbers (using while)
n=5
sum = 0
i =1
while i <=n:
    sum+=i
    i+=1

print("the total sum is =",sum)
```

```python
#WAP to find the factorial of first n numbers (using for)
n=3
fact= 1
i =1
while i <=n:
    fact*=i
    i+=1

print("the factorial is =",fact)

```

```python
#WAP to find the factorial of first n numbers(using for loop)
n=5
fact=1
for i in range(1,n+1):
    fact*=i

print("factorial =",fact)
```
