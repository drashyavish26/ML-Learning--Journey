```python
#lists in python = [ 1,3,4,4,] 
#MUTABLE 
marks=[93.3,34,36,87.6,86.7]
print(marks)
print(type(marks))
print(marks[0])
print(marks[4])
print(len(marks))
#can store data of diffrent  types 
student=["drashya ", 75 ," ichhawar" ]
print(student)
student[0]="abhi"#lists are mutable
print(student)
student[1]=85
print(student)
student[2]="indore"
print(student)
marks[1]=40
print(marks) #change in list (by indexing)
 


```

```python
#slicing in lists
marks=[93.3,34,36,87.6,86.7]
print(marks[1:5])
print(marks[:5])
print(marks[1:])
print(marks[-4:-3])
```

```python
#LIST METHODS = [1,3,4,5,]
#append or mutating the list (add elements to the list )
list=[ 2,1,3]
print(list)
list.append(4)
print(list)

#sorting- ascendeing desccending

'''list.sort ''' #sorts in ascending order 
'''list.sort(reverse=True)''' ##sorts in desccending order
'''list.reverse'''#sorts in reverse
(list.sort())
print(list)
(list.sort(reverse=True))

```

```python
#list methods 
list= ["b", "f", "a" , "c", "d", "e"]
print(list.sort())
print(list)
print(list.sort(reverse=True))  #after adding sort functions we can print in second line 
print(list)
```

```python
#insert method 
list= [2,1,3]
print(list)
list.append(4) # adds element to the last 
print(list)
print(list)
list.sort()#sorts in ascending order 
print(list)
list.sort(reverse=True)#sorts in descending order
print(list)
list.reverse()#reverses the list
print(list)
list.insert(1,5) #insert element st index 
print(list)
list.remove(1)#removes first occurences of element
print(list)
list.pop(1)#remove element at index
print(list)
```
