```python
#Nested dictionaries
student= {
"name": "drashya",
"subjects":{     #this is called nested in dict    
    "phy": 87,
    "chy":98,
     "mat":87,
}   # can make multiple dictionaries in dictionaries to store data
}
print(student)
print(student["subjects"])
print(student["subjects"]["chy"]) #can print data form nested dictionaries by using multiple brackets 

```

```python
#DIcTIONARIES METHOD 
student= {
"name": "drashya",
"subjects":{      
    "phy": 87,
    "chy":98,
     "mat":87,
}  
}
#keys method
print(student.keys())
print(list(student.keys()))
print(len(list(student.keys())))
print(len(list(student)))
print("   ")

#values  method
print(student.values()) #givess all values of all keys 
print(list(student.values())) # stores all values of all keys in a list
print("   ")


#items method
print(student.items()) #stores key and value pairs 
print(list(student.items()))
#[('name', 'drashya'), ('subjects', {'phy': 87, 'chy': 98, 'mat': 87})]
print(student.get("name"))
print(student.get("subjects"))
print(student.get("age")) #if key not present it gives none
print("   ")

#update method
(student.update({"city": "sehore"})) # can add more key values 
student.update({"name": " abhi"}) #changes update the value of keys 
print(student)





```

```python
set={ 1, 2, 3,4,5,5,}
set.add(6)
set.add("drashya") #we can add string , tuples in a set but not list only immutable objects are allowed 
set.remove(1)
print(set)
print(len(set))
set.pop() #removes a random value
set.pop()
print(set)
set.clear()
print(set)
set.add(2)
print(set)
print(len(set))
```

```python
#sets union(set2)& set.intersection(set2)
set={ 1, 2, 3,4,5,5,"drashya"}
set2={"drashya", "siya", "vishwakarma", 1 , 3 ,4}
set.union(set2)
print(set.union(set2)) #combines both sets
print(set.intersection(set2))#combines common values and returns them



```
